```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("../data/raw/itsm/all_tickets.csv")

print("Dataset Loaded:", df.shape)
```

```python
df["category"].value_counts().plot(
    kind="bar",
    figsize=(10, 5),
    title="Incident Category Distribution"
)

plt.xlabel("Category")
plt.ylabel("Number of Tickets")
plt.xticks(rotation=0)
plt.show()
```

```python
pd.crosstab(df["urgency"], df["impact"])
```

```python
pd.crosstab(df["category"], df["ticket_type"])
```

```python
pd.crosstab(
    df["category"],
    df["ticket_type"]
).plot(kind="bar", figsize=(12, 5))

plt.title("Ticket Type Distribution by Category")
plt.xlabel("Category")
plt.ylabel("Number of Tickets")
plt.xticks(rotation=0)
plt.show()
```

```python
corr = df.corr(numeric_only=True)

plt.figure(figsize=(10, 6))
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")

plt.title("Incident Data Correlation Heatmap")
plt.show()
```

## Key EDA Findings

Dataset contains 48,549 records and 9 columns.


712 missing values in title.


No duplicate records.


Category 4 accounts for approximately 70.2% of records.


Urgency and impact show a strong coded-value correlation (0.79).


Ticket type and urgency show a very high coded-value correlation (0.91).
