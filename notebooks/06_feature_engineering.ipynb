# Notebook 06 – Feature Engineering

## Objective
Create meaningful numerical and text-based features from IT incident data for machine learning models.

```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/processed/cleaned_it_incidents.csv")

print("Dataset Shape:", df.shape)
df.head()
```

```python
df["title_word_count"] = df["title"].str.split().str.len()
df["body_word_count"] = df["body"].str.split().str.len()

df[["title", "title_word_count", "body_word_count"]].head(10)
```

```python
df["has_password"] = df["body"].str.contains("password", case=False).astype(int)
df["has_network"] = df["body"].str.contains("network", case=False).astype(int)
df["has_email"] = df["body"].str.contains("email", case=False).astype(int)
df["has_access"] = df["body"].str.contains("access", case=False).astype(int)

df[["has_password", "has_network", "has_email", "has_access"]].head(10)
```

```python
df["title_length"] = df["title"].str.len()
df["body_length"] = df["body"].str.len()

df[["title_length", "body_length"]].describe()
```

```python
categorical_cols = [
    "ticket_type", "category", "sub_category1",
    "sub_category2", "business_service", "urgency", "impact"
]

for col in categorical_cols:
    df[col] = df[col].astype("category")

df[categorical_cols].dtypes
```

```python
df["title_missing"] = (df["title"].str.strip() == "").astype(int)

df["body_missing"] = (df["body"].str.strip() == "").astype(int)

df[["title_missing", "body_missing"]].sum()
```

```python
print("Final Dataset Shape:", df.shape)
print("Missing Values:", df.isnull().sum().sum())
print("Duplicate Rows:", df.duplicated().sum())

df.head()
```

```python
df.to_csv(
    "../data/processed/feature_engineered_incidents.csv",
    index=False
)

print("Feature Engineered Dataset Saved Successfully!")
```

# Feature Engineering – Conclusion

- Created title and body word-count features.
- Created keyword-based binary features.
- Created text-length features.
- Converted categorical codes into categorical data types.
- Created missing-text indicators.
- Exported the feature-engineered dataset.

The dataset is prepared for the next stage of machine learning experimentation.
