```python
import pandas as pd
import sqlite3
import matplotlib.pyplot as plt

print("OpsMind AI - SQL Analytics")
print("Libraries Imported Successfully!")
```

```python
df = pd.read_csv(
    "../data/processed/cleaned_it_incidents.csv"
)

print("Dataset Shape:", df.shape)
df.head()
```

```python
conn = sqlite3.connect(
    "../data/processed/opsmind_incidents.db"
)

print("SQLite Database Connected Successfully!")
```

```python
df.to_sql(
    "incidents",
    conn,
    if_exists="replace",
    index=False
)

print("Data Successfully Loaded into SQL!")
```

```python
query = """
SELECT * FROM incidents LIMIT 10;
"""

result = pd.read_sql_query(query, conn)
result
```

```python
query = """
SELECT * FROM incidents
WHERE urgency = 1;
"""

result = pd.read_sql_query(query, conn)
result.head(10)
```

```python
query = """
SELECT * FROM incidents
WHERE urgency = 1 AND impact = 3;
"""

result = pd.read_sql_query(query, conn)
result.head(10)
```

```python
query = """
SELECT * FROM incidents
WHERE urgency = 1 OR urgency = 2;
"""

result = pd.read_sql_query(query, conn)
result.head(10)
```

```python
query = """
SELECT DISTINCT category
FROM incidents;
"""

result = pd.read_sql_query(query, conn)
result
```

```python
query = """
SELECT COUNT(*) AS total_tickets
FROM incidents;
"""

result = pd.read_sql_query(query, conn)
result
```

```python
query = """
SELECT category, COUNT(*) AS total_tickets
FROM incidents
GROUP BY category
ORDER BY total_tickets DESC;
"""

result = pd.read_sql_query(query, conn)
result
```

```python
query = """
SELECT title, urgency, impact
FROM incidents
ORDER BY urgency ASC
LIMIT 10;
"""

result = pd.read_sql_query(query, conn)
result
```

```python
query = """
SELECT category, COUNT(*) AS total_tickets
FROM incidents
GROUP BY category
HAVING COUNT(*) > 1000;
"""

result = pd.read_sql_query(query, conn)
result
```

```python
query = """
SELECT urgency, impact, COUNT(*) AS total_tickets
FROM incidents
GROUP BY urgency, impact
ORDER BY total_tickets DESC;
"""

result = pd.read_sql_query(query, conn)
result
```
