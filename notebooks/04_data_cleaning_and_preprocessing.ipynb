```python
import pandas as pd
import numpy as np
import re

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)

print("OpsMind AI - Data Preprocessing")
print("Libraries Imported Successfully")
```

```python
df = pd.read_csv("../data/raw/itsm/all_tickets.csv")

print("Dataset Loaded Successfully")
print("Dataset Shape:", df.shape)
```

```python
missing_values = df.isnull().sum()
missing_values
```

```python
missing_percentage = (
    df.isnull().sum() / len(df) * 100
).round(2)

missing_percentage
```

```python
df["title"] = df["title"].fillna("")

print("Missing Titles After Cleaning:", df["title"].isnull().sum())
```

```python
duplicate_count = df.duplicated().sum()

print("Total Duplicate Records:", duplicate_count)
```

```python
df["title"].head(10)
```

```python
df["body"].head(5)
```

```python
print("Empty Titles:", (df["title"].str.strip() == "").sum())
print("Empty Bodies:", (df["body"].str.strip() == "").sum())
```

```python
df["title_missing"] = (
    df["title"].str.strip() == ""
).astype(int)

df["title_missing"].value_counts()
```

```python
df[["title", "title_missing", "body"]].head(10)
```

```python
df["title"] = df["title"].str.strip()
```

```python
df["body"] = df["body"].str.strip()
```

```python
print("Empty Titles:", (df["title"] == "").sum())
print("Empty Bodies:", (df["body"] == "").sum())
```

```python
df["title"] = df["title"].str.lower()
```

```python
df["body"] = df["body"].str.lower()
```

```python
df[["title", "body"]].head(10)
```

```python
df[df["title"].str.contains(
    r"[^a-zA-Z0-9\s]", na=False
)]["title"].head(10)
```

```python
df[df["body"].str.contains(
    r"[^a-zA-Z0-9\s]", na=False
)]["body"].head(10)
```

```python
df["title_length"] = df["title"].str.len()

df["title_length"].describe()
```

```python
df["body_length"] = df["body"].str.len()

df["body_length"].describe()
```

```python
df[["title", "title_length", "body", "body_length"]].head(10)
```

```python
df.nlargest(10, "body_length")[
    ["body_length", "body"]
]
```

```python
df.nsmallest(10, "body_length")[
    ["body_length", "body"]
]
```

```python
short_bodies = df[df["body_length"] <= 10]

print("Short Descriptions:", len(short_bodies))
print("Percentage:", round(len(short_bodies) / len(df) * 100, 2), "%")
```

```python
long_bodies = df[df["body_length"] > 1000]

print("Long Descriptions:", len(long_bodies))
print("Percentage:", round(len(long_bodies) / len(df) * 100, 2), "%")
```

## Objective

1. Loaded the IT incident dataset.

5. Created `title_missing` indicator.
6. Created `title_length` and `body_length` features.
7. Analyzed special characters and text-length distributions.
8. Validated categorical columns and their unique codes.
